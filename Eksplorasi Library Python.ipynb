#### **EKSPLORASI LIBRARY PYTHON UNTUK PROBABILITAS DAN DISTRIBUSI PROBABILITAS**

**1. math**

fungsi utama: `math.factorial(n)`, `math.perm(n, r)`, `math.comb(n, r)`, `math.exp(x)`, `math.log(x)`.

```python
import math
print("5!            =", math.factorial(5))
print("P(4,2)        =", math.perm(4, 2))
print("C(4,2)        =", math.comb(4, 2))
print("e^(-4)        =", round(math.exp(-4), 6))
```

**2. itertools**

fungsi utama: `product`, `permutations`, `combinations`, `combinations_with_replacement`.

```python
import itertools
print("product       :", list(itertools.product("AB", repeat=2)))
print("permutations  :", list(itertools.permutations("ABC", 2)))
print("combinations  :", list(itertools.combinations("ABC", 2)))
print("comb. w/ repl :", list(itertools.combinations_with_replacement("AB", 2)))
```

**3. random**

fungsi utama: `random.random()`, `random.choice()`, `random.sample()`, `random.randint()`, `random.seed()`.

```python
import random
random.seed(1)
print("angka acak 0-1  :", round(random.random(), 4))
print("pilih 1 server  :", random.choice(["S1", "S2", "S3"]))
print("sampel 2 dari 5 :", random.sample(range(1, 6), 2))
print("dadu (1-6)      :", random.randint(1, 6))
```

**5. collections.Counter**

fungsi utama: `Counter(iterable)` menghitung frekuensi tiap hasil.

```python
from collections import Counter
hasil = ["T", "H", "T", "T", "H"]
c = Counter(hasil)
print(c)
print("frekuensi relatif T =", c["T"] / len(hasil))
```

**9. scipy.stats**

library utama distribusi probabilitas. Setiap distribusi punya metode yang sama:

| Metode | Arti |
|---|---|
| `pmf(k)` | peluang tepat k (distribusi diskrit) |
| `pdf(x)` | kepadatan peluang (distribusi kontinu) |
| `cdf(x)` | peluang kumulatif P(X ≤ x) |
| `sf(x)` | P(X > x) = 1 - cdf(x) |
| `ppf(q)` | kebalikan cdf (nilai x untuk peluang kumulatif q) |
| `rvs(size)` | membangkitkan data acak |
| `stats(moments='mv')` | nilai harapan dan varians |

```python
from scipy import stats as st
X = st.binom(n=5, p=0.25)
print("P(X=3)       =", round(X.pmf(3), 4))
print("P(X<=2)      =", round(X.cdf(2), 4))
print("mean, var    =", X.stats(moments="mv"))
Z = st.norm(loc=200, scale=30)
print("P(X<250)     =", round(Z.cdf(250), 4))
print("persentil 95 =", round(Z.ppf(0.95), 2))
```

**10. matplotlib**

fungsi utama: `plt.bar`, `plt.plot`, `plt.hist`, `plt.fill_between`, `plt.show`.

```python
import matplotlib.pyplot as plt
k = np.arange(0, 6)
plt.bar(k, st.binom.pmf(k, 5, 0.25), color="steelblue")
plt.title("Contoh: pmf Binomial(n=5, p=0.25)")
plt.xlabel("k"); plt.ylabel("P(X=k)")
plt.show()
```

**11. sympy**

fungsi utama: `symbols`, `Rational`, `binomial`, `simplify`, `solve`.

```python
import sympy as sp
prev, se, fp = sp.symbols("prev se fp", positive=True)
posterior = prev * se / (prev * se + (1 - prev) * fp)
print("Rumus Bayes simbolik:", posterior)
print("C(4,2) simbolik     :", sp.binomial(4, 2))
```
